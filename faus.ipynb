# Histogramas y boxplots de variables cuantitativas

Según el diccionario (`diccionario_registros_enfr2018.xls`) las variables se dividen en:
* **Discretas**
* **Continuas** 
* **Categóricas codificadas**, el código `99` (Ns/Nc) se pasa a `NaN` antes de graficar.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Variables cuantitativas según el diccionario, agrupadas por bloque: {columna: descripción}
vars_discretas = {
    "Hogar y vivienda": {
        "bhcv02": "Ambientes de la vivienda",
        "bhho03": "Ambientes para dormir",
        "cant_componentes": "Miembros del hogar",
    },
    "Actividad física (días/semana)": {
        "biaf03": "Días act. física moderada",
        "biaf05": "Días que caminó ≥10 min",
    },
    "Alimentación y alcohol (días)": {
        "bial03": "Días/semana que come frutas",
        "bial05": "Días/semana que come verduras",
        "bica03_01": "Días/semana que tomó alcohol",
        "bica03_02": "Días/mes que tomó alcohol",
    },
    "Hipertensión": {
        "biha06": "Pastillas/día para la presión",
    },
}

vars_continuas = {
    "Ingresos y edad": {
        "bhih01": "Ingreso total mensual del hogar ($)",
        "bhch04": "Edad (años)",
    },
    "Actividad física (minutos)": {
        "biaf04_m": "Min/semana act. moderada",
        "biaf06_m": "Min/semana caminata",
        "biaf07_m": "Min/día sentado",
    },
    "Tabaco": {
        "bita02": "Edad en que fumó por primera vez",
    },
    "Alimentación (porciones)": {
        "bial04": "Porciones de fruta por día",
        "bial06": "Porciones de verdura por día",
        "promedio_fv_diario": "Promedio diario frutas y verduras",
    },
    "Mediciones antropométricas": {
        "promedio_sistolica": "Presión sistólica promedio (mm Hg)",
        "promedio_diastolica": "Presión diastólica promedio (mm Hg)",
        "bima07": "Altura medida (cm)",
        "bima10": "Peso medido (kg)",
        "imc_bima": "IMC medido",
        "bima13": "Perímetro de cintura (cm)",
    },
    "Mediciones bioquímicas": {
        "bimq05": "Glucemia (mg/dl)",
        "bimq06": "Colesterol total (mg/dl)",
    },
}
vars_codificadas = {
    "Ubicación": {
        "cod_provincia": "Código de provincia",
        "region": "Región (1 Metrop. 2 Pampeana 3 NOA 4 NEA 5 Cuyo 6 Patag.)",
        "tamanio_aglomerado": "Tamaño aglomerado (1 >1.5M ... 4 <150k hab.)",
    },
    "Ingresos y trabajo": {
        "rango_ingreso": "Rango de ingreso del hogar (0 a 17)",
        "bisl06": "Horas trabajadas (1 <35 / 2 35-45 / 3 >45)",
    },
    "Tabaco": {
        "bita03": "Fumó ≥100 cigarrillos (1 Sí / 2 No)",
        "bita04": "Fuma actualmente (1 Diario / 2 Algunos días / 3 No)",
        "bita05": "Marca de cigarrillos (código)",
    },
    "Mediciones (categorizadas)": {
        "imc_categorias_bima": "IMC medido (1 <25 / 2 25-30 / 3 >=30)",
        "colesterol_elevado": "Colesterol elevado (1 Sí / 2 No)",
        "glucemia_elevada": "Glucemia elevada (1 Sí / 2 No)",
    },
}

# Columnas donde 99 = Ns/Nc (no es un valor real)
cols_99_nsnc = ["rango_ingreso", "bisl06", "bita03", "imc_categorias_bima"]


def graficar_bloque(titulo, variables, discreta):
    """Una fila por variable: histograma a la izquierda y boxplot a la derecha."""
    n = len(variables)
    fig, axes = plt.subplots(n, 2, figsize=(14, 3.2 * n), squeeze=False,
                             gridspec_kw={"width_ratios": [3, 2]})
    for (ax_h, ax_b), (col, desc) in zip(axes, variables.items()):
        serie = raw_data[col]
        if col in cols_99_nsnc:
            serie = serie.replace(99, float("nan"))
        serie = serie.dropna()
        if discreta:
            sns.histplot(serie, discrete=True, ax=ax_h)
        else:
            sns.histplot(serie, bins=40, kde=True, ax=ax_h)
        ax_h.axvline(serie.mean(), color="red", ls="--", lw=1, label=f"media = {serie.mean():.1f}")
        ax_h.axvline(serie.median(), color="black", ls=":", lw=1, label=f"mediana = {serie.median():.1f}")
        ax_h.legend(fontsize=8)
        ax_h.set_title(f"{col} — {desc} (n = {len(serie)})")
        ax_h.set_xlabel("")

        sns.boxplot(x=serie, ax=ax_b, color="lightblue")
        ax_b.set_title(f"Boxplot {col}")
        ax_b.set_xlabel("")
    fig.suptitle(titulo, fontsize=14, y=1.0)
    plt.tight_layout()
    plt.show()


for titulo, variables in vars_discretas.items():
    graficar_bloque(f"{titulo} (discretas)", variables, discreta=True)

for titulo, variables in vars_continuas.items():
    graficar_bloque(f"{titulo} (continuas)", variables, discreta=False)

for titulo, variables in vars_codificadas.items():
    graficar_bloque(f"{titulo} (categóricas codificadas)", variables, discreta=True)
```

# Hipotesis NUESTRAS

## Hipotesis univariada

Todas las provincias estan representadas de manera distribuida correspondiente a la distribucion de Argentina 2018.

H0: La proporcion de la muestra por provincia es igual a su proporcion real.
H1: Al menos una de las provincia no esta correctamente representada.


### Muestra vs población por provincia

Para la población real se usa `wf1p`, el factor de expansión oficial del INDEC (Paso 1). Sumando `wf1p` por provincia se obtiene la población proyectada de 18 años y más de cada provincia en 2018 (≈ 27,9 millones de personas en total).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chisquare

provincias = {
    2: "Ciudad de Buenos Aires", 6: "Buenos Aires", 10: "Catamarca", 14: "Córdoba",
    18: "Corrientes", 22: "Chaco", 26: "Chubut", 30: "Entre Ríos", 34: "Formosa",
    38: "Jujuy", 42: "La Pampa", 46: "La Rioja", 50: "Mendoza", 54: "Misiones",
    58: "Neuquén", 62: "Río Negro", 66: "Salta", 70: "San Juan", 74: "San Luis",
    78: "Santa Cruz", 82: "Santa Fe", 86: "Santiago del Estero", 90: "Tucumán",
    94: "Tierra del Fuego",
}

df_prov = raw_data[["id", "cod_provincia", "wf1p"]].copy()
df_prov["provincia"] = df_prov["cod_provincia"].map(provincias)
assert df_prov["provincia"].notna().all(), "Hay códigos de provincia no reconocidos"

# Casos en la muestra vs población (suma de wf1p = población proyectada de 18+ por provincia, INDEC 2018)
resumen_provincia = df_prov.groupby("provincia").agg(
    casos=("id", "count"),
    poblacion=("wf1p", "sum"),
)
resumen_provincia["% muestra"] = resumen_provincia["casos"] / resumen_provincia["casos"].sum() * 100
resumen_provincia["% poblacion"] = resumen_provincia["poblacion"] / resumen_provincia["poblacion"].sum() * 100
resumen_provincia["diferencia (pp)"] = resumen_provincia["% muestra"] - resumen_provincia["% poblacion"]
resumen_provincia = resumen_provincia.sort_values("% poblacion", ascending=False)

print(f"Casos totales: {resumen_provincia['casos'].sum():,}")
print(f"Población total expandida (18+): {resumen_provincia['poblacion'].sum():,.0f}")
resumen_provincia.round(2)
```

```python
df_graf = resumen_provincia.sort_values("% poblacion")
y = np.arange(len(df_graf))
alto = 0.4

fig, ax = plt.subplots(figsize=(12, 10))
ax.barh(y - alto / 2, df_graf["% muestra"], alto, label="% en la muestra (casos)", color="#4C72B0")
ax.barh(y + alto / 2, df_graf["% poblacion"], alto, label="% en la población 2018 (Σ wf1p)", color="#55A868")
ax.set_yticks(y)
ax.set_yticklabels(df_graf.index)
ax.set_xlabel("Porcentaje (%)")
ax.set_title("Proporción por provincia: muestra vs población Argentina 2018")
ax.legend(fontsize=11)
ax.grid(axis="x", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

### Test chi-cuadrado de bondad de ajuste

Se compara la cantidad de casos observados por provincia con los esperados si la muestra fuera proporcional a la población: $E_i = n \cdot \pi_i$, con $\chi^2 = \sum \frac{(O_i - E_i)^2}{E_i}$ y $24 - 1 = 23$ grados de libertad.

Como con $n = 29.224$ casi cualquier diferencia resulta significativa, se reporta también el tamaño de efecto $w = \sqrt{\chi^2 / n}$ y los residuos estandarizados $(O_i - E_i)/\sqrt{E_i}$ para ver qué provincias se apartan más.

```python
# Test chi-cuadrado de bondad de ajuste
# Observados: casos por provincia. Esperados: n * proporción poblacional.
n = resumen_provincia["casos"].sum()
observados = resumen_provincia["casos"]
esperados = resumen_provincia["poblacion"] / resumen_provincia["poblacion"].sum() * n

chi2, p_valor = chisquare(f_obs=observados, f_exp=esperados)
gl = len(observados) - 1
w_cohen = np.sqrt(chi2 / n)  # tamaño de efecto: 0.1 chico, 0.3 medio, 0.5 grande

print(f"Chi² = {chi2:,.2f} | gl = {gl} | p-valor = {p_valor:.3f}")
print(f"w de Cohen = {w_cohen:.3f}")
print("Se rechaza H0" if p_valor < 0.05 else "No se rechaza H0", "(α = 0.05)")

# Residuos estandarizados: |r| > 2 indica qué provincias se apartan más de lo esperado
resumen_provincia["esperados"] = esperados
resumen_provincia["residuo"] = (observados - esperados) / np.sqrt(esperados)
resumen_provincia[["casos", "esperados", "residuo"]].sort_values("residuo").round(2)
```

### Conclusión

Se rechaza H0 ($\chi^2 \approx 18.376$, gl = 23, p < 0,001) con un tamaño de efecto grande ($w \approx 0{,}79$): **la muestra no es proporcional a la población por provincia**.

* Las provincias grandes están **subrepresentadas**: Buenos Aires (19% de la muestra vs 42% de la población), CABA, Santa Fe, Córdoba y Mendoza.
* Las provincias chicas están **sobrerrepresentadas**: Río Negro, Catamarca, Formosa, San Luis, La Rioja y La Pampa tienen entre 3 y 4 veces más casos de los esperados.
* Tucumán es la única con una proporción prácticamente igual a la poblacional.

Esto es esperable, porque la ENFR usa un muestreo estratificado que asigna un mínimo de casos a cada provincia para poder estimar a nivel provincial. Por eso, **para hacer estimaciones a nivel país hay que ponderar con `wf1p`** (`wf2p` y `wf3p` para las mediciones físicas y bioquímicas). Sin ponderar, los resultados quedarían sesgados hacia las provincias chicas.

## Hipotesis bivariada

El tamaño de la ciudad/aglomerado se asocia con los hábitos negativos para la salud (consumo de tabaco y consumo episódico excesivo de alcohol) del individuo.

* **$H_0$**: El tamaño del aglomerado y el patrón de consumo de hábitos nocivos (alcohol y tabaco) son independientes.
* **$H_1$**: El tamaño del aglomerado y el patrón de consumo de hábitos nocivos están asociados (la distribución de hábitos difiere según el tamaño de la ciudad).


### 1. Operacionalización y unificación de variables

Se usan variables oficiales de la ENFR 2018:
- **Tabaco** (`consumo_tabaco_100`): `Fumador actual` vs `No fumador` (los ex fumadores se agrupan con quienes nunca fumaron).
- **Alcohol** (`consumo_episodico_excesivo`): `Sí` / `No`. Se usa en lugar de "tomó en los últimos 30 días" porque esta última incluye consumo social moderado, que no es necesariamente un hábito nocivo.
- **Patrón unificado de hábitos**: `Ninguno`, `Solo tabaco`, `Solo alcohol excesivo`, `Ambos`.
- **Tamaño del aglomerado** (`tamanio_aglomerado`): `> 1.5M hab.`, `500k a 1.5M hab.`, `150k a 500k hab.` y `< 150k hab.`. Todos los casos de la categoría `> 1.5M` pertenecen a una única región (Gran Buenos Aires), por lo que esa categoría equivale a "GBA".

Se descartan los casos sin dato en alguna de las variables.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency

df_bi = raw_data[["id", "tamanio_aglomerado", "consumo_tabaco_100", "consumo_episodico_excesivo", "wf1p"]].copy()
df_bi = df_bi[
    df_bi["consumo_tabaco_100"].isin([1, 2, 3]) & df_bi["consumo_episodico_excesivo"].isin([1, 2])
].dropna(subset=["tamanio_aglomerado"]).copy()

#fumador actual (código 1) y consumo episódico excesivo (código 1)
df_bi["fuma"] = (df_bi["consumo_tabaco_100"] == 1).astype(int)
df_bi["toma"] = (df_bi["consumo_episodico_excesivo"] == 1).astype(int)

df_bi["patron_habito"] = np.select(
    [
        (df_bi["fuma"] == 0) & (df_bi["toma"] == 0),
        (df_bi["fuma"] == 1) & (df_bi["toma"] == 0),
        (df_bi["fuma"] == 0) & (df_bi["toma"] == 1),
    ],
    ["Ninguno", "Solo tabaco", "Solo alcohol excesivo"],
    default="Ambos",
)

# Tamaños de aglomerado según ENFR 2018
tam_etiquetas = {
    1: "> 1.5M hab. (GBA)",
    2: "500k - 1.5M hab.",
    3: "150k - 500k hab.",
    4: "< 150k hab. (Pequeñas)",
}
df_bi["tamanio_ciudad"] = df_bi["tamanio_aglomerado"].map(tam_etiquetas)
categorias_orden = list(tam_etiquetas.values())
habito_orden = ["Ninguno", "Solo tabaco", "Solo alcohol excesivo", "Ambos"]

# Tablas de contingencia observada y porcentajes fila (%)
tabla_obs = pd.crosstab(df_bi["tamanio_ciudad"], df_bi["patron_habito"]).reindex(index=categorias_orden, columns=habito_orden)
tabla_pct = pd.crosstab(df_bi["tamanio_ciudad"], df_bi["patron_habito"], normalize="index").reindex(index=categorias_orden, columns=habito_orden) * 100

print("--- Frecuencias Observadas (Casos) ---")
display(tabla_obs)
print("\n--- Proporciones Fila (%) ---")
display(tabla_pct.round(2))
```

### 2. Test Chi-cuadrado de Independencia y Tamaño del Efecto

Para contrastar la independencia entre dos variables cualitativas:
1. **Test $\chi^2$ de Independencia de Pearson** con frecuencias absolutas observadas ($O$) y esperadas ($E$).
2. **V de Cramér**: mide la magnitud o fuerza práctica de la asociación ($V \in [0, 1]$) controlando el tamaño de la muestra ($N \approx 29.000$).
3. **Residuos de Pearson**, para ver qué celdas aportan más al estadístico: valores mayores a $|2|$ indican desvíos importantes respecto de la independencia.

```python
# Test chi-cuadrado de independencia con frecuencias observadas
chi2_stat, p_val, dof, esperados = chi2_contingency(tabla_obs)

# Tamaño del efecto: V de Cramér
n_total = tabla_obs.values.sum()
r, k = tabla_obs.shape
v_cramer = np.sqrt(chi2_stat / (n_total * (min(r, k) - 1)))

# Residuos de Pearson y aporte de cada categoría de hábito al chi²
esperados_df = pd.DataFrame(esperados, index=tabla_obs.index, columns=tabla_obs.columns)
residuos = (tabla_obs - esperados_df) / np.sqrt(esperados_df)

print(f"Chi² = {chi2_stat:,.2f} | Grados de libertad = {dof} | p-valor = {p_val:.3e}")
print(f"Frecuencia esperada mínima = {esperados.min():.1f} (supuesto del chi²: todas > 5)")
print(f"V de Cramér = {v_cramer:.4f}")
print("\nResiduos de Pearson:")
display(residuos.round(2))
print("Aporte de cada hábito al chi²:")
display((residuos ** 2).sum().round(1).to_frame("aporte").T)
print("Resultado: Se rechaza H0 (hay asociación estadísticamente significativa)" if p_val < 0.05 else "No se rechaza H0")
```

```python
fig, ax = plt.subplots(figsize=(9, 6))

# Barras apiladas al 100% (composición relativa de hábitos por tamaño)
colores = ["#2ca02c", "#ff7f0e", "#1f77b4", "#d62728"]
tabla_pct.plot(kind="bar", stacked=True, ax=ax, color=colores, edgecolor="white", alpha=0.9)
ax.set_title("Composición de hábitos por tamaño de aglomerado (100%)", fontsize=12, pad=10)
ax.set_xlabel("Tamaño del aglomerado", fontsize=11)
ax.set_ylabel("Porcentaje (%)", fontsize=11)
ax.set_xticklabels(ax.get_xticklabels(), rotation=15, ha="right")
ax.legend(title="Patrón de consumo", bbox_to_anchor=(1.02, 1), loc="upper left")
ax.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# --- Comparación sin ponderar vs ponderado por wf1p ---
tabla_pond = pd.crosstab(
    df_bi["tamanio_ciudad"], df_bi["patron_habito"],
    values=df_bi["wf1p"], aggfunc="sum"
).reindex(index=categorias_orden, columns=habito_orden)
tabla_pct_pond = tabla_pond.div(tabla_pond.sum(axis=1), axis=0) * 100

print("% de la muestra vs % de la población por tamaño de aglomerado:")
display(pd.DataFrame({
    "% muestra": df_bi["tamanio_ciudad"].value_counts(normalize=True).reindex(categorias_orden) * 100,
    "% población": tabla_pond.sum(axis=1) / tabla_pond.sum().sum() * 100,
}).round(1))

print("\nProporciones fila (%) ponderadas por wf1p:")
display(tabla_pct_pond.round(2))

# --- Chequeo sin Gran Buenos Aires ---
# Todos los casos de la categoría > 1.5M son de una sola región, así que se verifica
# que la asociación no dependa solo de GBA.
tabla_sin_gba = tabla_obs.iloc[1:]
chi2_sg, p_sg, dof_sg, _ = chi2_contingency(tabla_sin_gba)
v_sg = np.sqrt(chi2_sg / (tabla_sin_gba.values.sum() * (min(tabla_sin_gba.shape) - 1)))
print(f"\nSin GBA: Chi² = {chi2_sg:,.2f} | gl = {dof_sg} | p-valor = {p_sg:.3e} | V de Cramér = {v_sg:.4f}")
```

### 3. ¿Aporta el tabaco? Comparación con cada hábito por separado

Para ver cuánto aporta cada hábito, se repite el test con alcohol solo (sin tabaco) y con tabaco solo, sobre la misma muestra.

```python
resultados = {}
for nombre, col in [("Alcohol excesivo", "toma"), ("Tabaco", "fuma")]:
    tab = pd.crosstab(df_bi["tamanio_ciudad"], df_bi[col]).reindex(index=categorias_orden, columns=[0, 1])
    chi2_h, p_h, dof_h, _ = chi2_contingency(tab)
    v_h = np.sqrt(chi2_h / tab.values.sum())  # min(r, k) - 1 = 1
    pct_h = tab[1] / tab.sum(axis=1) * 100
    pond_h = pd.crosstab(df_bi["tamanio_ciudad"], df_bi[col], values=df_bi["wf1p"], aggfunc="sum").reindex(index=categorias_orden)
    pct_pond_h = pond_h[1] / pond_h.sum(axis=1) * 100
    odds = (pct_h.iloc[0] / (100 - pct_h.iloc[0])) / (pct_h.iloc[-1] / (100 - pct_h.iloc[-1]))
    resultados[nombre] = (pct_h, pct_pond_h)
    print(f"{nombre}: Chi² = {chi2_h:,.2f} | gl = {dof_h} | p-valor = {p_h:.3e} | V de Cramér = {v_h:.4f} | OR (> 1.5M vs < 150k) = {odds:.2f}")

fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)
x = np.arange(len(categorias_orden))
ancho = 0.38
for ax, (nombre, (pct_h, pct_pond_h)) in zip(axes, resultados.items()):
    ax.bar(x - ancho / 2, pct_h.values, ancho, label="Sin ponderar", color="#1f77b4", edgecolor="white")
    ax.bar(x + ancho / 2, pct_pond_h.values, ancho, label="Ponderado (wf1p)", color="#ff7f0e", edgecolor="white")
    ax.set_title(f"{nombre} según tamaño de aglomerado", fontsize=12, pad=10)
    ax.set_xlabel("Tamaño del aglomerado", fontsize=11)
    ax.set_xticks(x)
    ax.set_xticklabels(categorias_orden, rotation=15, ha="right")
    ax.grid(axis="y", linestyle="--", alpha=0.5)
axes[0].set_ylabel("% de personas", fontsize=11)
axes[0].legend()

plt.tight_layout()
plt.show()
```

#### Lectura de estos dos gráficos

**Lo que importa es la forma de las barras, no su altura.** La hipótesis pregunta si el hábito cambia según el tamaño de la ciudad, así que hay que ver cómo varían las barras de izquierda a derecha. La altura solo indica qué tan común es el hábito en general: fuma ~22% de la gente y toma en exceso ~12%, por eso las barras del tabaco son más altas en todas las ciudades. Eso no significa que el tabaco dependa más del tamaño de la ciudad.

| | > 1.5M (GBA) | 500k - 1.5M | 150k - 500k | < 150k | V de Cramér | OR (GBA vs < 150k) |
|---|---|---|---|---|---|---|
| Alcohol excesivo | 13,8% | 12,5% | 12,7% | 10,3% | 0,043 | 1,40 |
| Tabaco | 21,6% | 24,3% | 21,4% | 20,7% | 0,026 | 1,05 |

- **Alcohol excesivo:** las barras bajan de las ciudades grandes a las chicas. Hay un patrón claro con el tamaño.
- **Tabaco:** las barras suben y bajan sin orden, y la primera y la última son casi iguales. No hay un patrón con el tamaño.

Por eso, aunque el tabaco tenga barras más altas, aporta menos a la hipótesis.

### 4. Conclusiones del análisis bivariado

1. **Resultado del test $\chi^2$**: Se rechaza la hipótesis nula $H_0$ ($\chi^2 \approx 69{,}95$, $\text{gl} = 9$, $p \approx 1{,}6 \times 10^{-11}$). Existe una asociación estadísticamente significativa entre el tamaño del aglomerado y el patrón de hábitos nocivos.
2. **Tamaño del efecto (V de Cramér $\approx 0{,}028$)**: La asociación es **muy débil**. El test es significativo por el gran volumen muestral ($N \approx 29.000$), pero el tamaño de la ciudad explica una parte mínima de los hábitos.
3. **Tendencias observadas en los datos**:
   - **Alcohol excesivo exclusivo**: es del **8,3%** en GBA y del **6,2%** en localidades de menos de 150k habitantes.
   - **Ambos hábitos**: 5,5% en GBA contra 4,1% en las localidades chicas.
   - **Ninguno**: 70,1% en GBA contra 73,1% en las localidades chicas.
   - **Tabaco exclusivo**: se mantiene entre 16,1% y 18,7% sin una tendencia clara.
   - Los residuos de Pearson más grandes corresponden a las categorías con alcohol, en los extremos de tamaño (GBA con más y localidades chicas con menos de lo esperado).
4. **El tabaco aporta poco**: Por separado, el alcohol excesivo muestra una asociación mayor ($V \approx 0{,}043$; OR $\approx 1{,}40$ entre GBA y localidades chicas) que el tabaco ($V \approx 0{,}026$; OR $\approx 1{,}05$). El alcohol excesivo desciende de forma ordenada del 13,8% en GBA al 10,3% en las localidades chicas, es decir, es ~34% más frecuente en GBA. El tabaco, en cambio, no sigue una tendencia con el tamaño: sube al 24,3% en las ciudades de 500k a 1,5M, y es prácticamente igual en GBA (21,6%) y en las localidades chicas (20,7%), apenas ~4% más frecuente en GBA. Al sumar el tabaco al patrón unificado, el $\chi^2$ pasa de 54,3 (solo alcohol) a 70,0, un aumento modesto aun con 6 grados de libertad adicionales: el tabaco aporta poco a la asociación con el tamaño de la ciudad.

5. **Limitaciones**:
   - La categoría `> 1.5M` coincide con una única región (GBA), por lo que el efecto "metrópolis" es en realidad "GBA contra el resto". Excluyendo GBA la asociación se mantiene pero más débil ($\chi^2 \approx 47{,}66$, gl = 6, $p \approx 1{,}4 \times 10^{-8}$, $V \approx 0{,}031$).
   - La muestra sobrerrepresenta a las ciudades chicas (58% de los casos contra 30% de la población). Con las proporciones ponderadas por `wf1p` las diferencias se mantienen e incluso se acentúan levemente.
   - La ENFR tiene un diseño muestral complejo (estratificado y por conglomerados) y el chi² supone muestreo aleatorio simple, por lo que los p-valores son algo optimistas. Con p tan chicos no cambia la decisión.
   - Con una definición más amplia de alcohol ("consumió en los últimos 30 días") la asociación es más fuerte, pero esa variable incluye consumo moderado.

**Conclusión final**: Se rechaza $H_0$. El tamaño del aglomerado se asocia con los hábitos nocivos, principalmente con el consumo episódico excesivo de alcohol, que es más frecuente en las ciudades grandes (en particular GBA) que en las pequeñas. El tabaco no muestra una tendencia clara con el tamaño de la ciudad y aporta poco a la asociación del conjunto, que resulta muy débil.

## Hipotesis multivariada

Los hábitos, como una alimentación completa, actividad física regular, un IMC controlado y un consumo bajo de alcohol y tabaco, están relacionados con las principales enfermedades (hipertensión y diabetes).

H0: Los hábitos no están relacionados con la presencia de hipertensión o diabetes.
H1: Los hábitos están relacionados con la presencia de hipertensión o diabetes.


### Clusters con UMAP

Se toma una muestra de 3.000 casos con edad, sexo, tabaco, alcohol e IMC, se estandarizan y se reducen a dos dimensiones con UMAP. El mismo mapa se colorea por cada variable y por la enfermedad (hipertensión o diabetes) para ver si los agrupamientos que forman los hábitos coinciden con dónde se concentran los enfermos.


```python
import umap
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# 1. Preparar datos
df_demo = raw_data[['id', 'bhch04', 'bhch03']].copy()
df_demo['edad'] = pd.to_numeric(df_demo['bhch04'], errors='coerce')
df_demo['sexo'] = np.where(pd.to_numeric(df_demo['bhch03'], errors='coerce') == 1, 'Hombre', 'Mujer')

df_tabaco = raw_data[['id', 'consumo_tabaco_100']].copy()
df_tabaco['fuma'] = np.where(df_tabaco['consumo_tabaco_100'].isin([1, 2]), 'Sí', 'No')

df_alcohol = raw_data[['id', 'bica01']].copy()
df_alcohol['toma_alcohol'] = np.where(df_alcohol['bica01'] == 1, 'Sí', 'No')

df_peso = raw_data[['id', 'imc']].copy()

df_salud = raw_data[['id', 'prevalencia_hipertension', 'prevalencia_diabetes']].copy()
df_salud['enfermedad'] = np.where((df_salud['prevalencia_hipertension'] == 1) | (df_salud['prevalencia_diabetes'] == 1), 'Sí', 'No')

# Merge de todas las variables
df = df_demo[['id', 'edad', 'sexo']].merge(df_tabaco[['id', 'fuma']], on='id')\
                                    .merge(df_alcohol[['id', 'toma_alcohol']], on='id')\
                                    .merge(df_peso, on='id')\
                                    .merge(df_salud[['id', 'enfermedad']], on='id').dropna()

# 2. Muestra representativa para procesar rápidamente (puedes aumentar n si tienes buena RAM)
np.random.seed(42)
df_sample = df.sample(n=3000, random_state=42)

# 3. Binarizar variables categóricas para el modelo matemático
X = df_sample[['edad', 'imc']].copy()
X['sexo_num'] = np.where(df_sample['sexo'] == 'Hombre', 1, 0)
X['fuma_num'] = np.where(df_sample['fuma'] == 'Sí', 1, 0)
X['alcohol_num'] = np.where(df_sample['toma_alcohol'] == 'Sí', 1, 0)

# Estandarizar escalas
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Ejecutar UMAP
reducer = umap.UMAP(n_neighbors=30, min_dist=0.3, metric='euclidean', random_state=42)
embedding = reducer.fit_transform(X_scaled)
df_sample['umap_1'] = embedding[:, 0]
df_sample['umap_2'] = embedding[:, 1]

# 5. Generar la cuadrícula visual de mapeo
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

# Parámetros visuales base
plot_kws = {'s': 15, 'alpha': 0.8, 'edgecolor': None}

# Edad
sns.scatterplot(x='umap_1', y='umap_2', hue='edad', palette='viridis', data=df_sample, ax=axes[0], **plot_kws)
axes[0].set_title('Coloreado por EDAD')

# Sexo
sns.scatterplot(x='umap_1', y='umap_2', hue='sexo', palette='Set2', data=df_sample, ax=axes[1], **plot_kws)
axes[1].set_title('Coloreado por SEXO')

# Tabaco
sns.scatterplot(x='umap_1', y='umap_2', hue='fuma', palette='Reds', data=df_sample, ax=axes[2], **plot_kws)
axes[2].set_title('Coloreado por TABACO')

# Alcohol
sns.scatterplot(x='umap_1', y='umap_2', hue='toma_alcohol', palette='Blues', data=df_sample, ax=axes[3], **plot_kws)
axes[3].set_title('Coloreado por ALCOHOL')

# IMC
sns.scatterplot(x='umap_1', y='umap_2', hue='imc', palette='magma', data=df_sample, ax=axes[4], **plot_kws)
axes[4].set_title('Coloreado por IMC')

# Enfermedad
sns.scatterplot(x='umap_1', y='umap_2', hue='enfermedad', palette={'Sí': '#d62728', 'No': '#e0e0e0'}, data=df_sample, ax=axes[5], **plot_kws)
axes[5].set_title('Distribución de la ENFERMEDAD (HTA/Diabetes)')

# Limpiar los ejes para estética
for ax in axes:
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_xlabel('')
    ax.set_ylabel('')

plt.tight_layout()
plt.show()
```

# Hipotesis GRUPAL

## Hipotesis univariada



" El consumo diario de frutas y verduras en la población adulta argentina
es inferior a las 5 porciones diarias recomendadas por la OMS. "

H0 : Consumo diario >= 5
H1 : Consumo diario < 5

```python
df_hipotesis_univariada_grupal = raw_data["promedio_fv_diario"].copy()

df_hipotesis_univariada_grupal.describe()
```

### Normalización Min-Max

Se lleva `promedio_fv_diario` a la escala $[0, 1]$.

```python
# Normalización Min-Max: x' = (x - min) / (max - min)  ->  escala [0, 1]
fv = df_hipotesis_univariada_grupal.dropna()
fv_min, fv_max = fv.min(), fv.max()

fv_minmax = (fv - fv_min) / (fv_max - fv_min)
# Referencia OMS (5 porciones) llevada a la misma escala
ref_minmax = (5 - fv_min) / (fv_max - fv_min)

print(f"min = {fv_min}, max = {fv_max}")
fv_minmax.describe()
```

### Normalidad: Shapiro-Wilk y QQ-plot

H0: los datos siguen una distribución normal. | H1: no siguen una distribución normal.

```python
from scipy import stats

# Shapiro-Wilk. H0: los datos provienen de una distribución normal
# (OJO n > 5000)
w_shapiro, p_shapiro = stats.shapiro(fv_minmax)
print(f"Shapiro-Wilk: W = {w_shapiro:.4f} | p-valor = {p_shapiro:.3g}")
print("Se rechaza normalidad" if p_shapiro < 0.05 else "No se rechaza normalidad", "(α = 0.05)")
print(f"Asimetría mediante skew = {stats.skew(fv_minmax):.2f}")

fig, (ax_h, ax_qq) = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(fv_minmax, bins=40, kde=True, ax=ax_h)
ax_h.axvline(ref_minmax, color="red", ls="--", label=f"Referencia OMS (5 porciones = {ref_minmax:.3f})")
ax_h.axvline(fv_minmax.mean(), color="black", ls=":", label=f"Media = {fv_minmax.mean():.3f}")
ax_h.set_title("Distribución de promedio_fv_diario (Min-Max)")
ax_h.set_xlabel("Porciones diarias normalizadas")
ax_h.legend()

stats.probplot(fv_minmax, dist="norm", plot=ax_qq)
ax_qq.set_title("QQ-plot vs distribución normal")

plt.tight_layout()
plt.show()
```

### Test t de Student (una muestra, unilateral)

H0: $\mu \geq 5$ porciones | H1: $\mu < 5$ porciones (en escala Min-Max, $\mu_0 = 5/18 \approx 0{,}278$).

```python
# Test t de Student de una muestra, unilateral a izquierda
t_stat, p_t = stats.ttest_1samp(fv_minmax, popmean=ref_minmax, alternative="less")
gl_t = len(fv_minmax) - 1

print(f"Media muestral (Min-Max) = {fv_minmax.mean():.4f} vs referencia = {ref_minmax:.4f}")
print(f"t = {t_stat:.2f} | gl = {gl_t} | p-valor = {p_t:.3f}")
print("Se rechaza H0: el consumo es menor a 5 porciones" if p_t < 0.05 else "No se rechaza H0", "(α = 0.05)")

# Intervalo de confianza unilateral 95% (cota superior), devuelto a porciones
ic_sup = fv_minmax.mean() + stats.t.ppf(0.95, gl_t) * fv_minmax.std(ddof=1) / np.sqrt(len(fv_minmax))
print(f"Cota superior IC 95%: {ic_sup:.4f} (Min-Max) = {ic_sup * (fv_max - fv_min) + fv_min:.2f} porciones")
```

### Sin ponderar vs ponderado por `wf1p`

Como se vio en el chi-cuadrado por provincia, la muestra no es proporcional a la población. Para concluir sobre la *población adulta argentina* se pondera cada caso por `wf1p`. Para el test t ponderado se usa el tamaño efectivo de Kish, $n_{ef} = \frac{(\sum w)^2}{\sum w^2}$, que refleja la pérdida de precisión por los pesos desiguales.

```python
# Comparación sin ponderar vs ponderado por wf1p (factor de expansión Paso 1)
pesos = raw_data.loc[fv_minmax.index, "wf1p"]


def resumen_t(x, w, mu0):
    """Media, desvío y test t unilateral (H1: μ < mu0) con pesos.
    Con pesos se usa el tamaño efectivo de Kish: n_ef = (Σw)² / Σw²."""
    media = np.average(x, weights=w)
    desvio = np.sqrt(np.average((x - media) ** 2, weights=w))
    n_ef = w.sum() ** 2 / (w ** 2).sum()
    t = (media - mu0) / (desvio / np.sqrt(n_ef))
    p = stats.t.cdf(t, df=n_ef - 1)
    return {
        "media (Min-Max)": media,
        "media (porciones)": media * (fv_max - fv_min) + fv_min,
        "desvío (Min-Max)": desvio,
        "% con >= 5 porciones": np.average(x >= ref_minmax, weights=w) * 100,
        "n (efectivo)": n_ef,
        "t": t,
        "p-valor": p,
    }


comparacion = pd.DataFrame({
    "Sin ponderar": resumen_t(fv_minmax, np.ones(len(fv_minmax)), ref_minmax),
    "Ponderado (wf1p)": resumen_t(fv_minmax, pesos, ref_minmax),
})
comparacion["Diferencia (pond - sin pond)"] = comparacion["Ponderado (wf1p)"] - comparacion["Sin ponderar"]
comparacion.round(4)
```

### Conclusión

* **Normalidad:** Shapiro-Wilk rechaza la normalidad (W ≈ 0,797, p < 0,001). El QQ-plot muestra una asimetría positiva fuerte (asimetría ≈ 2,5): la mayoría consume pocas porciones y hay pocos casos con valores muy altos. El test t se justifica igual por el TCL.
* **Test t:** se rechaza H0 (t ≈ −289, p < 0,001). El consumo medio es de **2,07 porciones diarias** sin ponderar, y la cota superior del IC 95% es 2,09 porciones, muy por debajo de las 5 recomendadas por la OMS.
* **Ponderado vs sin ponderar:** al ponderar por `wf1p`, la media baja a **2,03 porciones** (−0,04) y la proporción que llega a 5 porciones o más pasa de 6,2% a **6,0%**. El n efectivo cae de 28.471 a ≈ 8.994 por la desigualdad de los pesos, y el t pasa a ≈ −171. **La conclusión no cambia: el consumo de frutas y verduras de la población adulta argentina es inferior a 5 porciones diarias**, y solo alrededor de 6 de cada 100 adultos cumple la recomendación.
