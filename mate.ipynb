```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar datos originales
raw_data = pd.read_csv('ENFR 2018 - Base usuario.txt', sep='|')

# 1. Recrear las 'tablas apartadas' relevantes del código de entrega
df_ingresos = raw_data[["id", "bhih01", "bhih01_02", "bhih03", "rango_ingreso", "quintil_uc", "imputado"]].copy()

df_componentes = raw_data[["id", "bhch03_j", "bhch04_j", "bhch05_j", "bhch10_01_j", "bhch10_02_j",
                           "bhch10_03_j", "bhch10_04_j", "bhch10_05_j", "bhch10_06_j", "bhch10_99_j",
                           "bhch02", "bhch03", "bhch04", "bhch05", "bhch10_01", "bhch10_02", "bhch10_03",
                           "bhch10_04", "bhch10_05", "bhch10_06", "bhch10_99", "cant_componentes",
                           "miembros_18", "rango_edad_j", "nivel_instruccion_j", "nivel_instruccion_agrupado_j",
                           "cobertura_salud_j"]].copy()

raw_data_actividad_fisica = raw_data[["id", "biaf01", "biaf02_m", "biaf02_99", "biaf03", "biaf04_m",
                                      "biaf04_99", "biaf05", "biaf06_m", "biaf06_99", "biaf07_m",
                                      "biaf07_99", "biaf08", "biaf09", "biaf10_01", "biaf10_02",
                                      "biaf10_03", "biaf10_04", "nivel_actividad_fisica", "barreras_actividad_fisica"]].copy()

raw_data_peso_corporal = raw_data[["id", "bipc01", "bipc02", "bipc03", "bipc04", "bipc04_99",
                                   "bipc05", "bipc05_99", "imc", "imc_categorias"]].copy()

raw_data_mediciones_antro = raw_data[["id", "bima01", "bima02", "bima03", "bima04_01_a", "bima04_01_b",
                                      "bima04_02_a", "bima04_02_b", "bima04_03_a", "bima04_03_b",
                                      "promedio_sistolica", "promedio_diastolica", "ta_elevada",
                                      "prevalencia_hipertension_combinada", "bima06", "bima07",
                                      "bima09", "bima10", "imc_bima", "imc_categorias_bima",
                                      "bima12", "bima13", "bima14"]].copy()

raw_data_alimentacion = raw_data[["id", "bial01", "bial02", "bial03", "bial03_99", "bial04",
                                  "bial04_99", "bial05", "bial05_99", "bial06", "bial06_99", "bial07",
                                  "bial08", "bial09", "bial10", "promedio_fv_diario", "consumo_fv",
                                  "barreras_fyv", "tipo_dieta_razones"]].copy()

# 2. Hacer los joins (merge) usando 'id' como clave para armar los datasets de análisis
# Dataset para histogramas
df_hist = df_ingresos[['id', 'bhih01']].merge(df_componentes[['id', 'bhch04_j']], on='id')
df_hist = df_hist.merge(raw_data_peso_corporal[['id', 'imc', 'bipc04', 'bipc05']], on='id')
df_hist = df_hist.merge(raw_data_actividad_fisica[['id', 'biaf06_m']], on='id')

# Dataset para matriz de correlación antropométrica
df_antro = raw_data_peso_corporal[['id', 'bipc04', 'bipc05', 'imc']].merge(
    raw_data_mediciones_antro[['id', 'bima04_01_a', 'bima04_01_b', 'promedio_sistolica', 'promedio_diastolica']],
    on='id')

# Dataset para matriz de correlación de hábitos
df_habitos = raw_data_actividad_fisica[['id', 'biaf02_m', 'biaf04_m', 'biaf06_m', 'biaf07_m']].merge(
    raw_data_alimentacion[['id', 'promedio_fv_diario']],
    on='id')

# 3. Generar Gráficos
# Histogramas
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()
cols_hist = ['bhih01', 'bhch04_j', 'imc', 'bipc04', 'bipc05', 'biaf06_m']
nombres_hist = ['Ingreso del Hogar', 'Edad Jefe/a', 'IMC', 'Peso Reportado (kg)', 'Altura Reportada (cm)', 'Minutos Caminata']

for i, col in enumerate(cols_hist):
    sns.histplot(df_hist[col].dropna(), bins=30, ax=axes[i], kde=True, color='mediumpurple')
    axes[i].set_title(f'Distribución de {nombres_hist[i]}')
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Frecuencia')

plt.tight_layout()
plt.show()

# Matriz Correlación Antropometría
plt.figure(figsize=(8, 6))
sns.heatmap(df_antro.drop(columns=['id']).corr(), annot=True, cmap='viridis', fmt=".2f", vmin=-1, vmax=1)
plt.title('Matriz de Correlación: Variables Antropométricas y Presión')
plt.tight_layout()
plt.show()

# Matriz Correlación Hábitos
plt.figure(figsize=(8, 6))
sns.heatmap(df_habitos.drop(columns=['id']).corr(), annot=True, cmap='viridis', fmt=".2f", vmin=-1, vmax=1)
plt.title('Matriz de Correlación: Hábitos (Actividad y Alimentación)')
plt.tight_layout()
plt.show()
```

## Distribución de Variables Clave

Las distribuciones muestran el comportamiento general de la población encuestada en factores sociodemográficos y de salud.
- Ingreso del Hogar y Minutos de Caminata: Presentan una fuerte asimetría positiva (sesgo a la derecha), lo que indica que la mayoría de los encuestados reporta valores bajos, con una larga cola de valores atípicos altos.
- Edad del Jefe/a de Hogar: Muestra una distribución más uniforme, con una ligera concentración en los rangos de 40 a 60 años.
- IMC y Peso: Presentan distribuciones en forma de campana pero con un sesgo hacia la derecha, reflejando la prevalencia de sobrepeso y obesidad en la muestra

## Distribución de Variables Clave (Merged Datasets)
Los histogramas se generaron uniendo df_ingresos, df_componentes, raw_data_peso_corporal y raw_data_actividad_fisica. Las distribuciones resultantes confirman el mismo comportamiento observado previamente en la base general: el ingreso y la actividad física presentan una fuerte asimetría positiva, mientras que el IMC y el peso muestran un sesgo a la derecha.

### Análisis de Correlación: Variables Antropométricas y de Presión
Esta matriz se construyó cruzando raw_data_peso_corporal y raw_data_mediciones_antro. Se mantiene la alta redundancia estructural entre las mediciones físicas. El peso y el IMC mantienen su colinealidad crítica de 0.84, y las diferentes tomas de presión arterial sistólica y diastólica superan el umbral de 0.80 con sus respectivos promedios, confirmando que se debe utilizar un único indicador para evitar ruido.

#### Análisis de Correlación: Hábitos y Actividad Física
Al cruzar raw_data_actividad_fisica y raw_data_alimentacion, los resultados reafirman la independencia lineal entre los distintos hábitos de salud. Las correlaciones entre el consumo de frutas y verduras, los minutos de actividad moderada o vigorosa, y el tiempo de sedentarismo se mantienen cercanas a 0, permitiendo incluir estas variables en un mismo modelo multivariado sin riesgo de multicolinealidad.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Creación de variables binarias
raw_data_tabaco['fuma_actualmente'] = np.where(raw_data_tabaco['consumo_tabaco_100'].isin([1, 2]), 1, 0)

raw_data_alcohol['toma_alcohol'] = np.where(raw_data_alcohol['bica01'] == 1, 1, 0)
raw_data_alcohol.loc[raw_data_alcohol['bica01'] == 99, 'toma_alcohol'] = np.nan

raw_data_alcohol['alcohol_excesivo'] = np.where(raw_data_alcohol['consumo_episodico_excesivo'] == 1, 1, 0)
raw_data_alcohol.loc[raw_data_alcohol['consumo_episodico_excesivo'].isna(), 'alcohol_excesivo'] = np.nan

# 2. CORRECCIÓN: Unir ambos datasets por su Índice en lugar de la columna 'id'
df_vicios = raw_data_tabaco[['fuma_actualmente']].merge(
    raw_data_alcohol[['toma_alcohol', 'alcohol_excesivo']],
    left_index=True,
    right_index=True
).dropna()

# 3. Clasificación de perfiles de consumo para el gráfico legible
condiciones = [
    (df_vicios['fuma_actualmente'] == 1) & (df_vicios['toma_alcohol'] == 1),
    (df_vicios['fuma_actualmente'] == 0) & (df_vicios['toma_alcohol'] == 1),
    (df_vicios['fuma_actualmente'] == 1) & (df_vicios['toma_alcohol'] == 0),
    (df_vicios['fuma_actualmente'] == 0) & (df_vicios['toma_alcohol'] == 0)
]
opciones = ['Ambos (Fuma y Toma)', 'Solo Toma Alcohol', 'Solo Fuma', 'Ninguno']
# 3. Clasificación de perfiles de consumo para el gráfico legible
condiciones = [
    (df_vicios['fuma_actualmente'] == 1) & (df_vicios['toma_alcohol'] == 1),
    (df_vicios['fuma_actualmente'] == 0) & (df_vicios['toma_alcohol'] == 1),
    (df_vicios['fuma_actualmente'] == 1) & (df_vicios['toma_alcohol'] == 0),
    (df_vicios['fuma_actualmente'] == 0) & (df_vicios['toma_alcohol'] == 0)
]
opciones = ['Ambos (Fuma y Toma)', 'Solo Toma Alcohol', 'Solo Fuma', 'Ninguno']

# CORRECCIÓN: Agregar default='Desconocido' para que todos los datos sean tipo string
df_vicios['perfil_consumo'] = np.select(condiciones, opciones, default='Desconocido')

# 4. Gráfico 1: Prevalencia General
plt.figure(figsize=(8, 5))
prevalencias = df_vicios[['fuma_actualmente', 'toma_alcohol', 'alcohol_excesivo']].mean() * 100
sns.barplot(x=prevalencias.index, y=prevalencias.values, palette='Reds')
plt.title('Prevalencia en la Muestra: Tabaco y Alcohol (%)')
plt.ylabel('Porcentaje (%)')
plt.xticks(ticks=[0, 1, 2], labels=['Fuma Actualmente', 'Tomó Alcohol (30 días)', 'Alcohol Excesivo (CEE)'])
plt.ylim(0, 100)
for i, v in enumerate(prevalencias):
    plt.text(i, v + 1, f"{v:.1f}%", ha='center', fontweight='bold')
plt.tight_layout()
plt.show()

# 5. Gráfico 2: Correlación Lineal
plt.figure(figsize=(6, 5))
corr_matrix = df_vicios[['fuma_actualmente', 'toma_alcohol', 'alcohol_excesivo']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1)
plt.title('Correlación entre Tabaco y Alcohol')
plt.xticks(ticks=[0.5, 1.5, 2.5], labels=['Fuma', 'Alcohol (30d)', 'CEE'], rotation=0)
plt.yticks(ticks=[0.5, 1.5, 2.5], labels=['Fuma', 'Alcohol (30d)', 'CEE'], rotation=0)
plt.tight_layout()
plt.show()

# 6. Gráfico 3: Intersección de Perfiles (Quién fuma y quién toma)
plt.figure(figsize=(9, 6))
perfiles = df_vicios['perfil_consumo'].value_counts(normalize=True) * 100
orden = ['Ninguno', 'Solo Fuma', 'Solo Toma Alcohol', 'Ambos (Fuma y Toma)']
perfiles = perfiles.reindex(orden)

sns.barplot(x=perfiles.index, y=perfiles.values, palette='magma')
plt.title('Perfil de Consumo: Intersección de Tabaco y Alcohol')
plt.ylabel('Porcentaje de la Muestra (%)')
plt.ylim(0, 100)
for i, v in enumerate(perfiles.values):
    plt.text(i, v + 1.5, f"{v:.1f}%", ha='center', fontweight='bold', fontsize=11)
plt.tight_layout()
plt.show()
```

## Prevalencia de Consumo
A partir de los datos declarados, observamos que el 78.8% de los encuestados reporta haber consumido alcohol en los últimos 30 días, estableciéndose como el hábito más extendido. En contraparte, un 41.1% de la muestra sin ponderar se identifica como fumador actual (diario u ocasional), mientras que el consumo episódico excesivo de alcohol (CEE) tiene una prevalencia mucho menor, ubicándose en un 11.4%.

## Correlación entre Hábitos
Al analizar si estos comportamientos ocurren en simultáneo en las mismas personas, la matriz de correlación nos muestra una relación positiva pero débil (0.19) entre el consumo de tabaco y el de alcohol. De manera similar, la correlación entre ser fumador y reportar consumo excesivo de alcohol (CEE) es aún más baja (0.12). Esto sugiere que, a nivel estadístico general, el consumo de una sustancia no es un predictor fuerte de la otra en esta población.

## Intersección: ¿Los que beben fuman más?
A pesar de la baja correlación lineal global, el análisis cruzado demuestra que la prevalencia de tabaquismo es significativamente superior entre los bebedores.

* Entre quienes sí consumieron alcohol en el último mes, el 45.9% también fuma.

* Entre quienes no consumieron alcohol, el porcentaje de fumadores desciende al 23.4%.

Por lo tanto, aunque no hay una colinealidad fuerte que justifique fusionar las variables, existe una clara tendencia de agrupación: es casi el doble de probable que un encuestado fume si también reporta ingesta reciente de alcohol.

# Hipótesis bivariada: IMC según nivel de actividad física

## Inciso
> *El índice de masa corporal, calculado a partir de la medición física de peso y talla (del paso 2), cambia según el nivel de actividad física de las personas.*

## Hipótesis
- **H0:** la distribución del IMC medido es la misma en los tres niveles de actividad física (Alta, Media, Baja).
- **H1:** el IMC medido difiere entre al menos dos niveles de actividad física. Lo esperado es que a menor actividad, mayor IMC.

## Variables utilizadas
| Variable | Tipo | Rol | Descripción |
|---|---|---|---|
| `imc_bima` | Cuantitativa continua | **Respuesta** | IMC (kg/m²) calculado con el peso y la talla **medidos** en el paso 2 |
| `nivel_actividad_fisica` | Cualitativa ordinal (1 = Alta, 2 = Media, 3 = Baja) | **Explicativa** | Agrupa a las personas en 3 niveles de actividad física |


## Preparación de los datos
Solo una submuestra de personas pasó por el paso 2 (mediciones físicas), así que se descartan los registros sin `imc_bima`. También se excluyen los códigos 99 (NS/NC).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from itertools import combinations

raw_data = pd.read_csv('ENFR 2018 - Base usuario.txt', sep='|')

# Variables: IMC (paso 2) y nivel de actividad física (1=Alta, 2=Media, 3=Baja)
df_h = raw_data[['id', 'imc_bima', 'imc_categorias_bima', 'nivel_actividad_fisica']].copy()
df_h = df_h.dropna(subset=['imc_bima', 'nivel_actividad_fisica'])
df_h = df_h[df_h['imc_categorias_bima'] != 99]

df_h['actividad'] = df_h['nivel_actividad_fisica'].map({1: 'Alta', 2: 'Media', 3: 'Baja'})
orden = ['Alta', 'Media', 'Baja']
colores = {'Alta': '#2a9d8f', 'Media': '#e9c46a', 'Baja': '#e76f51'}

print(f"n total = {len(df_h)}\n")
print(df_h.groupby('actividad')['imc_bima'].describe().loc[orden].round(2))
```

**Interpretación:** después del filtrado quedan alrededor de 16.300 personas, y el grupo más grande es el de actividad **Baja**. Tanto la media como la mediana del IMC **suben a medida que baja la actividad física**, y el desvío estándar también es mayor en el grupo Baja. Esto ya anticipa lo que hay que testear: si los grupos difieren y si sus varianzas son iguales.

### Boxplot del IMC por nivel de actividad

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_h, x='actividad', y='imc_bima', order=orden, hue='actividad', palette=colores, legend=False, showfliers=False)
plt.title('IMC medido según nivel de actividad física')
plt.xlabel('Nivel de actividad física')
plt.ylabel('IMC (kg/m²)')
plt.show()
```

**Interpretación:** la caja (rango intercuartílico) y la mediana se desplazan hacia arriba de Alta a Baja. Los bigotes superiores son más largos que los inferiores, lo que indica **asimetría a la derecha**: hay personas con IMC muy alto que estiran la distribución. Los grupos se superponen bastante, así que la diferencia existe pero no es enorme.

## Elección del test: verificación de supuestos
Para comparar una variable continua entre 3 grupos, la opción paramétrica es el **ANOVA de un factor**, que exige dos supuestos:
1. **Normalidad** del IMC dentro de cada grupo.
2. **Homocedasticidad**: varianzas iguales entre los grupos.

Primero se mira la forma de las distribuciones con histogramas y Q-Q plots.

```python
bins = np.arange(12, 62, 1)
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
for i, g in enumerate(orden):
    x = df_h.loc[df_h['actividad'] == g, 'imc_bima']
    mu, sd = x.mean(), x.std()

    # Primeros graficos : histograma para observar normalidad de las variables
    ax = axes[0, i]
    ax.hist(x, bins=bins, density=True, color=colores[g], alpha=0.75, edgecolor='white')
    xs = np.linspace(12, 62, 300)
    ax.plot(xs, stats.norm.pdf(xs, mu, sd), 'k--', lw=1.5, label='Normal teórica')
    ax.axvline(x.median(), color='k', lw=1, label=f'Mediana = {x.median():.1f}')
    ax.set_title(f'Actividad {g} (n = {len(x)})\nasimetría = {stats.skew(x):.2f}, curtosis = {stats.kurtosis(x):.2f}')
    ax.set_xlabel('IMC (kg/m²)')
    ax.set_ylabel('Densidad')
    ax.legend(fontsize=8)

    # Segundos Graficos: QQ Plot para observar distribucion de normalidad
    stats.probplot(x, dist='norm', plot=axes[1, i])
    axes[1, i].get_lines()[0].set(markerfacecolor=colores[g], markeredgecolor=colores[g], markersize=3, alpha=0.5)
    axes[1, i].set_title(f'Q-Q plot – Actividad {g}')

plt.suptitle('Distribución del IMC medido según nivel de actividad física', fontsize=14)
plt.tight_layout()
plt.show()
```

**Interpretación:**
- **Histogramas:** en los tres grupos el pico queda a la izquierda de la normal teórica y hay una cola larga hacia valores altos de IMC. La asimetría es positiva (cerca de 1) y la curtosis es mayor que 0, así que las colas son más pesadas que las de una normal.
- **Q-Q plots:** si el IMC fuera normal, los puntos caerían sobre la recta roja. En cambio, forman una curva: en el extremo derecho se van por encima de la recta (valores más altos de lo esperado) y en el izquierdo también se separan.

Visualmente, **el IMC no parece normal en ningún grupo**. Se confirma con un test formal.

### Supuesto 1: normalidad (test de Kolmogorov-Smirnov)
- **H0:** el IMC del grupo sigue una distribución normal.
- **H1:** el IMC del grupo no sigue una distribución normal.

Se compara la distribución de cada grupo con una normal que tiene su misma media y su mismo desvío. El nivel de significación es α = 0.05.

```python
filas = []
for g in orden:
    x = df_h.loc[df_h['actividad'] == g, 'imc_bima'].values
    D, p_ks = stats.kstest(x, stats.norm(loc=x.mean(), scale=x.std(ddof=1)).cdf)
    filas.append({'Actividad': g, 'n': len(x), 'Estadístico D': round(D, 4),
                  'p-valor': f'{p_ks:.2e}', '¿Normal? (α=0.05)': 'Sí' if p_ks > 0.05 else 'No'})
normalidad = pd.DataFrame(filas).set_index('Actividad')
normalidad
```

**Interpretación:** en los tres grupos el p-valor es muchísimo menor a 0.05, así que **se rechaza H0**: el IMC **no sigue una distribución normal** en ningún nivel de actividad. Esto coincide con lo que mostraban los histogramas y los Q-Q plots. Con muestras tan grandes el test detecta hasta desvíos chicos, pero acá la asimetría también se ve a simple vista, así que la conclusión es firme.

**No se cumple el supuesto de normalidad.**

### Supuesto 2: homocedasticidad (test de Levene centrado en la mediana)
- **H0:** las varianzas del IMC son iguales en los tres niveles de actividad.
- **H1:** al menos una de las varianzas es distinta.

Se usa la versión de Levene centrada en la **mediana** (Brown-Forsythe), porque es más robusta cuando los datos no son normales, como en este caso.

```python
grupos = [df_h.loc[df_h['actividad'] == g, 'imc_bima'] for g in orden]

print('Varianza del IMC por grupo:')
print(df_h.groupby('actividad')['imc_bima'].var().loc[orden].round(2))

W, p_lev = stats.levene(*grupos, center='median')
print(f"\nLevene (mediana): W = {W:.2f}, p-valor = {p_lev:.2e}")
print('¿Varianzas iguales? (α=0.05):', 'Sí' if p_lev > 0.05 else 'No')
```

**Interpretación:** el p-valor es mucho menor a 0.05, así que **se rechaza H0**: las varianzas del IMC **no son iguales** entre los grupos. La varianza crece a medida que baja la actividad física, por lo que el grupo Baja es el más disperso.

**No se cumple el supuesto de homocedasticidad.**

## Test Elegido
No se cumple **ninguno** de los dos supuestos del ANOVA, así que **no corresponde usar ANOVA**. Se usa su alternativa no paramétrica, el **test de Kruskal-Wallis**:

### Test de Kruskal-Wallis
- **H0:** la distribución del IMC es la misma en los tres niveles de actividad física.
- **H1:** en al menos un nivel de actividad el IMC tiende a tomar valores distintos que en otro.

Además del p-valor se calcula el **tamaño de efecto ε²** (épsilon cuadrado), que mide qué tan grande es la diferencia. Hace falta porque con n tan grande casi cualquier diferencia da significativa.

```python
H, p_kw = stats.kruskal(*grupos)
gl = len(orden) - 1
eps2 = H / (len(df_h) - 1)

print(f"Kruskal-Wallis: H = {H:.2f}, gl = {gl}, p-valor = {p_kw:.2e}")
print(f"Tamaño de efecto: epsilon² = {eps2:.3f}")
print('¿Se rechaza H0? (α=0.05):', 'Sí' if p_kw < 0.05 else 'No')
```

**Interpretación:** el p-valor es prácticamente 0 (muy por debajo de 0.05), así que **se rechaza H0**: el IMC medido **difiere según el nivel de actividad física**.

Sin embargo, el **ε² es chico** (alrededor de 0.02): menos del 2% de la variabilidad en los rangos del IMC se asocia al nivel de actividad. La diferencia es estadísticamente clara pero de **magnitud modesta**, porque el IMC depende de muchos otros factores.

### Comparaciones de a pares (Mann-Whitney con corrección de Bonferroni)
Kruskal-Wallis dice que *al menos* un grupo difiere, pero no cuál. Para saberlo se compara cada par de grupos con Mann-Whitney. Como se hacen 3 comparaciones, se aplica la **corrección de Bonferroni**: cada p-valor se multiplica por 3 para no inflar el error tipo I.

```python
n_comp = len(list(combinations(orden, 2)))
filas = []
for a, b in combinations(orden, 2):
    xa = df_h.loc[df_h['actividad'] == a, 'imc_bima']
    xb = df_h.loc[df_h['actividad'] == b, 'imc_bima']
    U, p_ab = stats.mannwhitneyu(xa, xb)
    p_adj = min(p_ab * n_comp, 1)
    filas.append({'Comparación': f'{a} vs {b}',
                  'Mediana 1': round(xa.median(), 2), 'Mediana 2': round(xb.median(), 2),
                  'p ajustado': f'{p_adj:.2e}', '¿Difieren? (α=0.05)': 'Sí' if p_adj < 0.05 else 'No'})
pd.DataFrame(filas).set_index('Comparación')
```

**Interpretación:** los tres pares dan p ajustado < 0.05, así que **los tres niveles de actividad difieren entre sí**. El orden de las medianas coincide con lo esperado: **Alta < Media < Baja**. A menor actividad física, mayor IMC.

## Conclusión
- **Se rechaza H0**: hay evidencia estadística de que el IMC medido **cambia según el nivel de actividad física**, y tiende a ser mayor cuanto menor es la actividad.
- Se usó **Kruskal-Wallis en lugar de ANOVA** porque el IMC no es normal (Kolmogorov-Smirnov) y las varianzas no son iguales (Levene con mediana).
- **El efecto es chico** (ε² ≈ 0.02): con una muestra tan grande la diferencia sale muy significativa, pero la actividad física explica una parte pequeña de la variación del IMC.
- Como las varianzas difieren entre grupos, el resultado se interpreta como que en los grupos con menos actividad el IMC **tiende a tomar valores más altos**, no estrictamente como una diferencia de medianas.
- **Limitaciones:** es una asociación bivariada y **no prueba causalidad**. La relación podría ir en sentido inverso (un IMC alto puede llevar a hacer menos actividad), y hay posibles variables confusoras como la **edad**, que se asocia tanto con el IMC como con la actividad física.
