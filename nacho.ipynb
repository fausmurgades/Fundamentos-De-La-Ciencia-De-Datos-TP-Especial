```python
import pandas as pd

raw_data = pd.read_csv('ENFR 2018 - Base usuario.txt', sep = '|')
df = raw_data.copy()

# Obtenemos todas las columnas que empiezan con BH
cols_BH = df.columns[df.columns.str.startswith('bh')].tolist()

# Separamos en subgrupos
cols_BHCV = df.columns[df.columns.str.startswith('bhcv')].tolist() # Vivienda
cols_BHHO = df.columns[df.columns.str.startswith('bhho')].tolist() # Hogar
cols_BHIH = df.columns[df.columns.str.startswith('bhih')].tolist() # Ingresos
cols_BHCH = df.columns[df.columns.str.startswith('bhch')].tolist() # Componentes
cols_BHSL = df.columns[df.columns.str.startswith('bhsl')].tolist() # Situacion Laboral Jefe 

# Identificador y Ubicacion
cols_ID_UB = ['id', 'cod_provincia', 'region', 'tamanio_aglomerado', 'aglomerado', 'localidades_150', 'submuestra']

# Añadimos a Características del Hogar
cols_BHHO.extend(['tipo_hogar'])

# Añadimos a Ingresos del Hogar
cols_BHIH.extend(['rango_ingreso', 'quintil_uc'])

# Añadimos a Componentes del Hogar 
cols_BHCH.extend([
    'cant_componentes', 'miembros_18', 'rango_edad_j', 
    'nivel_instruccion_j', 'nivel_instruccion_agrupado_j', 'cobertura_salud_j'
])

# Añadimos a Situación Laboral del Jefe del Hogar
cols_BHSL.extend(['condicion_actividad_j'])

cols_agrupadas_bh = [col for col in cols_BH if col in (cols_BHCV + cols_BHHO + cols_BHIH + cols_BHCH + cols_BHSL)]
cols_afuera = [col for col in cols_BH if col not in cols_agrupadas_bh]

# Diccionario 
categorias_columnas = {
    "Identificación y Ubicación (ID/UB)": cols_ID_UB,
    "Características de la Vivienda (BHCV)": cols_BHCV,
    "Características del Hogar (BHHO)": cols_BHHO,
    "Ingresos del Hogar (BHIH)": cols_BHIH,
    "Componentes del hogar (BHCH)": cols_BHCH,
    "Situacion Laboral del Jefe del Hogar": cols_BHSL,
    "SOBRANTES (BH sin clasificar)": cols_afuera
}
# Print
for nombre_categoria, columnas in categorias_columnas.items():
    print(f"--- {nombre_categoria} ---")
    print(f"Cantidad encontradas: {len(columnas)}")
    print(f"Columnas: {columnas}")
    print()
```
