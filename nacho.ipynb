```python
df_id_ub_copia=raw_data[['id', 'cod_provincia', 'wf1p']].copy()

provincias = {
    2: "Ciudad de Buenos Aires", 6: "Buenos Aires", 10: "Catamarca", 14: "Córdoba",
    18: "Corrientes", 22: "Chaco", 26: "Chubut", 30: "Entre Ríos", 34: "Formosa",
    38: "Jujuy", 42: "La Pampa", 46: "La Rioja", 50: "Mendoza", 54: "Misiones",
    58: "Neuquén", 62: "Río Negro", 66: "Salta", 70: "San Juan", 74: "San Luis",
    78: "Santa Cruz", 82: "Santa Fe", 86: "Santiago del Estero", 90: "Tucumán",
    94: "Tierra del Fuego",
}

df_id_ub_copia['wf1p']=df_id_ub_copia['wf1p'].astype(str).str.replace(',','.').astype(float)
df_id_ub_copia['wf1p'].isna().sum()
```

```python
df_id_ub_copia['provincia']=df_id_ub_copia['cod_provincia'].map(provincias)
print(df_id_ub_copia)

# Validar que todas las provincias hayan sido mapeadas correctamente
assert df_id_ub_copia['provincia'].isna().sum() == 0, "Existen códigos de provincia no reconocidos"
```

```python
resumen_provincia = df_id_ub_copia['provincia'].value_counts().rename('Casos').to_frame()
resumen_provincia['% muestra'] = (resumen_provincia['Casos'] / resumen_provincia['Casos'].sum() * 100).round(2)
resumen_provincia['Poblacion']=df_id_ub_copia.groupby('provincia')['wf1p'].sum()
resumen_provincia['% pob'] = (resumen_provincia['Poblacion'] / resumen_provincia['Poblacion'].sum() * 100).round(2)

resumen_provincia
```

NOTE

¿Por qué wf1p? wf1p es el factor de expansión/ponderación muestral oficial del INDEC. Al sumar los valores de wf1p por provincia, obtienes exactamente la población proyectada de 18 años y más de cada provincia en 2018 (un universo de ~27,9 millones de personas).

```python
df_comp = pd.DataFrame({
    '% Muestra Encuestada': resumen_provincia['% muestra'],
    '% Población Real 2018': resumen_provincia['% pob']
}).sort_values(by='% Población Real 2018', ascending=True)

df_comp['Diferencia (% Muestra - % Pob)'] = df_comp['% Muestra Encuestada'] - df_comp['% Población Real 2018']

df_comp
```

```python
import numpy as np
import matplotlib.pyplot as plt

y = np.arange(len(df_comp))
height = 0.4
fig, ax = plt.subplots(figsize=(12, 10))
ax.barh(y - height/2, df_comp['% Muestra Encuestada'], height, label='% en la Muestra (Encuestas)', color='#4C72B0')
ax.barh(y + height/2, df_comp['% Población Real 2018'], height, label='% en Población Real 2018 (INDEC)', color='#55A868')
ax.set_yticks(y)
ax.set_yticklabels(df_comp.index)
ax.set_xlabel('Porcentaje (%)')
ax.set_title('Comparación: Proporción en la Muestra vs Población Real 2018 por Provincia')
ax.legend(fontsize=11)
ax.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

df_comp.sort_values(by='% Población Real 2018', ascending=False)
```

```python
from scipy.stats import shapiro

for col in ['% Muestra Encuestada', '% Población Real 2018', 'Diferencia (% Muestra - % Pob)']:
    stat, p = shapiro(df_comp[col])
    print(f"{col}: W = {stat:.4f}, p = {p:.4f}")

# Al tener un p valor menor a 0.05, se rechaza la hipotesis nula y no se distribuye de manera normal
```

```python
from scipy.stats import levene

stat, p = levene(df_comp['% Muestra Encuestada'], df_comp['% Población Real 2018'])
print(f"Levene: W = {stat:.4f}, p = {p:.4f}")

# p valor es menor que 0.05, asi que tampoco es homocedastica
```

```python
from scipy.stats import kruskal

stat, p = kruskal(df_comp['% Muestra Encuestada'], df_comp['% Población Real 2018'])
print(f"Estadístico={stat:.3f}, p-valor={p:.3f}")

# Hay diferencia significativa entre la proporcion de encuestas respondidas por provincia y su composicion a nivel pais.
```
